## What Feature Importance Means

Decision Tree feature importance estimates how much each feature helped reduce impurity in the tree splits. A larger value means the feature was more useful to this fitted model.

Feature importance does not prove causation. It only describes how this model used the available features.

## Why Decision Trees Are Interpretable

A Decision Tree predicts by following a path of feature-based rules. This makes it easier to explain than many black-box models, especially when the tree is not too large.
